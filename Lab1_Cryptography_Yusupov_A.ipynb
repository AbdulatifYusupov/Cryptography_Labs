## Affine Cipher
$C = (a \cdot p + b) \bmod n$

```python
from math import gcd

def count_affine_keys(n):
    phi = sum(1 for i in range(1, n) if gcd(i, n) == 1)
    return phi * n

print("English (26):", count_affine_keys(26))
print("Uzbek (29):  ", count_affine_keys(29))
print("Slovak (46): ", count_affine_keys(46))
```

## Polygraphic substitution cipher (length 4)

## Affine cipher decryption

**Ciphertext:** `NEJFX VPCOB PYUKN RUKPC TPUBP VKBO`  
**Known plaintext:** first two letters = `if`


```python
from string import ascii_uppercase

alphabet = ascii_uppercase
a_inv, b = 9, 15

ciphertext = "NEJFX VPCOB PYUKN RUKPC TPUBP VKBO"
plaintext = ""

for char in ciphertext:
    if char.isalpha():
        C = alphabet.index(char)
        P = (a_inv * (C - b)) % 26
        plaintext += alphabet[P]
    else:
        plaintext += char

print(plaintext)
```

# Decipher (monoalphabetic substitution cipher):

```python
def atbash_decode(text: str) -> str:
    result = []
    
    en_lower = "abcdefghijklmnopqrstuvwxyz"
    en_upper = en_lower.upper()
    
    for char in text:
        if char in en_lower:  # Заменено elif на if
            idx = en_lower.index(char)
            result.append(en_lower[-1 - idx])
        elif char in en_upper:
            idx = en_upper.index(char)
            result.append(en_upper[-1 - idx])
        else:
            result.append(char)
            
    return "".join(result)

cipher_text = input()
decoded_text = atbash_decode(cipher_text)
print(decoded_text)

#Zgyzhs rh z hrnkov hfyhgrgfgrlm xrksvi uli gsv Svyivd zokszyvg.
```
